````python
# Tests des API Endpoints

print("\n🔌 TESTS DES ENDPOINTS API")
print("=" * 60)

# Test Authentification - Inscription
print("\n📝 Test 2.1 : Inscription (API)")
try:
    payload = {
        "email": f"test_user_{datetime.now().timestamp()}@test.com",
        "password": "Test123!@#",
        "firstName": "Test",
        "lastName": "User"
    }
    response = requests.post(f"{BASE_URL}/api/auth/register", json=payload)
    if response.status_code in [200, 201]:
        print(f"✅ Inscription réussie (200/201)")
        user_data = response.json()
        print(f"   Email: {user_data.get('email', 'N/A')}")
    elif response.status_code == 400:
        print(f"⚠️  Erreur 400 - Données invalides: {response.json().get('message', 'N/A')}")
    else:
        print(f"❌ Erreur HTTP {response.status_code}")
except Exception as e:
    print(f"❌ Erreur: {e}")

# Test Authentification - Connexion
print("\n📝 Test 2.2 : Connexion (API)")
try:
    payload = {
        "email": "acheteur@test.com",
        "password": "password123"
    }
    response = requests.post(f"{BASE_URL}/api/auth/login", json=payload)
    if response.status_code == 200:
        print(f"✅ Connexion réussie (200)")
        data = response.json()
        token = data.get('token')
        user_id = data.get('user', {}).get('id')
        print(f"   Token obtenu: {token[:20] if token else 'N/A'}...")
        print(f"   User ID: {user_id}")
    elif response.status_code == 401:
        print(f"⚠️  Identifiants invalides (401)")
    else:
        print(f"❌ Erreur HTTP {response.status_code}")
except Exception as e:
    print(f"⚠️  API Auth non disponible ou erreur: {e}")

# Test Matching
print("\n📝 Test 1.4 : Matching API")
try:
    params = {
        "budget_min": 100000,
        "budget_max": 300000,
        "surface_min": 50,
        "limit": 5
    }
    response = requests.get(f"{BASE_URL}/api/matching", params=params)
    if response.status_code == 200:
        print(f"✅ Matching API accessible (200)")
        results = response.json()
        annonces_count = len(results) if isinstance(results, list) else results.get('count', 0)
        print(f"   Nombre d'annonces trouvées: {annonces_count}")
    else:
        print(f"⚠️  Code HTTP: {response.status_code}")
except Exception as e:
    print(f"⚠️  API Matching erreur: {e}")

# Test Annonces
print("\n📝 Test 1.4 (suite) : Annonces API")
try:
    response = requests.get(f"{BASE_URL}/api/annonces")
    if response.status_code == 200:
        print(f"✅ API Annonces accessible (200)")
        annonces = response.json()
        annonces_count = len(annonces) if isinstance(annonces, list) else annonces.get('total', 0)
        print(f"   Nombre d'annonces en base: {annonces_count}")
    else:
        print(f"⚠️  Code HTTP: {response.status_code}")
except Exception as e:
    print(f"⚠️  API Annonces erreur: {e}")

# Test FAQ
print("\n📝 Test FAQ API")
try:
    response = requests.get(f"{BASE_URL}/api/faq")
    if response.status_code == 200:
        print(f"✅ FAQ API accessible (200)")
        faqs = response.json()
        faq_count = len(faqs) if isinstance(faqs, list) else faqs.get('total', 0)
        print(f"   Nombre de FAQs: {faq_count}")
    else:
        print(f"⚠️  Code HTTP: {response.status_code}")
except Exception as e:
    print(f"⚠️  API FAQ erreur: {e}")

# Test Chatbot
print("\n📝 Test Chatbot API")
try:
    payload = {"message": "Comment estimer mon bien ?"}
    response = requests.post(f"{BASE_URL}/api/chatbot", json=payload)
    if response.status_code == 200:
        print(f"✅ API Chatbot accessible (200)")
        chatbot_response = response.json()
        print(f"   Réponse: {chatbot_response.get('response', 'N/A')[:50]}...")
    else:
        print(f"⚠️  Code HTTP: {response.status_code}")
except Exception as e:
    print(f"⚠️  API Chatbot erreur: {e}")

print("\n" + "=" * 60)
````
