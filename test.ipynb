Instructions pour les Tests

🔹 Préparation

Backend : Lance ton serveur Flask (python app.py).
Frontend : Ouvre les pages HTML dans un navigateur (ex: file:///chemin/vers/index.html ou via Flask).
Données de test :

Crée des comptes test (acheteur, vendeur, acheteur+vendeur).
Ajoute des annonces fictives en base de données.

Outils :

Chrome DevTools (F12) pour inspecter les requêtes API et les erreurs.
Postman ou curl pour tester les endpoints API manuellement.

🔹 Exécution des Tests

Suis les étapes décrites dans chaque test.
Note le statut (🟢/🟡/🔴/⏳) et les commentaires.
Documente les bugs dans la section "Suivi des Bugs".
🔹 Validation

Un test est validé (🟢) si le résultat attendu correspond au résultat réel.
Un test est en échec (🔴) si le résultat réel ne correspond pas ou si une erreur survient.

📌 6. Checklist avant Déploiement

 Tous les tests critiques (⭐⭐⭐) sont validés (🟢).
 Aucun bug critique (🔴) n’est présent dans la section "Suivi des Bugs".
 Les workflows complets fonctionnent sans erreur.
 Les messages d’erreur sont clairs et aident l’utilisateur.
 Le site est testé sur mobile (responsive).
 Le site est testé sur plusieurs navigateurs (Chrome, Firefox, Edge).
